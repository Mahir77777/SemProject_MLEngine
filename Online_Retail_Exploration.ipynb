**1. Datasets considered**

Our project, StockMate, will predict next week’s product sales to help shop owners decide what to restock.

| Dataset                                                                                                    | What it contains                                    | What we could predict                                                                                         |
| ---------------------------------------------------------------------------------------------------------- | --------------------------------------------------- | ------------------------------------------------------------------------------------------------------------- |
| [Online Retail](https://archive.ics.uci.edu/dataset/352/online+retail)                                     | Product details, quantities sold and sale dates.    | Weekly sales quantity for each product using `Quantity`.                                                      |
| [Online Retail II](https://archive.ics.uci.edu/dataset/502/online+retail+ii)                               | Similar retail data covering two years.             | Weekly sales quantity for each product using `Quantity`.                                                      |
| [Daily Demand Forecasting Orders](https://archive.ics.uci.edu/dataset/409/daily+demand+forecasting+orders) | Daily orders from a logistics company over 60 days. | Total daily orders using `Target_(Total_orders)`. This is less suited to predicting individual product sales. |


**2. Selected dataset**

We chose Online Retail because it has the product codes, quantities and dates needed for StockMate. It is also smaller and easier to work with than Online Retail II.

We will add up the quantities sold for each product each week, after handling returns and cancellations. The model will use this history to predict next week’s sales.

The data comes from a UK retailer, so we would need to test the model with Kenyan shop data before using it locally. Shop owners will enter their current stock in the app.
